```python
from tcgdexsdk import TCGdex, Language, Query

tcgdex = TCGdex("en")

```

```python
async def get_card(card_name, tcgdex=tcgdex):

    card_variants = await tcgdex.card.list(Query().equal("name", card_name)) # Gets a list of every card matching the card name

    if card_variants is None:
        return f"Could not find any cards named {card_name}"

    return await tcgdex.card.get(card_variants[0].id) # Returns the first entry of the list based on that entry's card ID (e.g. sv09-97)
```

```python
parsed_decklist = dict()
card_info = dict()

with open ("decklist.txt") as decklist:
    for line in decklist:
        line = line.strip()

        if not line or line.split(":", 1)[0] in {"Pokémon", "Trainer", "Energy"}:
            continue

        parts = line.split()
        
        count = int(parts[0])
        card_name = " ".join(parts[1:-2]) 


        parsed_decklist[card_name] = count

        card = await get_card(card_name)
        card_info[card_name] = {"stage": card.stage, "hp": card.hp, "type": card.types, "suffix": card.suffix}

        
```
